# Plotting Backends

skrf plots with [matplotlib](https://matplotlib.org/) by default, and can also draw interactive [plotly](https://plotly.com/python/) figures. Plotly is optional: install it with `pip install scikit-rf[plotly]`.

This notebook draws each kind of plot with both backends.

```python
%matplotlib inline
import plotly.io as pio

import skrf as rf
from skrf.data import ring_slot, ring_slot_meas

pio.renderers.default = 'plotly_mimetype+notebook_connected'  # also show plotly figures in the HTML docs
```

## Selecting a backend

Set the backend for all plots with `rf.options.plotting.backend`, which is `'matplotlib'` or `'plotly'`. With plotly, the plot methods return a `plotly.graph_objects.Figure`.

```python
rf.options.plotting.backend = 'plotly'
ring_slot.plot_s_db()
```

A single call can choose a backend with `backend=`:

```python
ring_slot.plot_s_db(backend='matplotlib')
```

```python
rf.options.plotting.backend = 'matplotlib'  # back to the default
```

## Overlays

Matplotlib draws on the current axes, as before. With plotly there is no current figure: pass the figure as `ax` to add traces to it. Frequencies are converted to the unit of the figure's first trace.

```python
fig = ring_slot.plot_s_db(m=0, n=0, label='Theory', backend='plotly')
ring_slot_meas.plot_s_db(m=0, n=0, label='Measurement', ax=fig)
```

## Styling

Common matplotlib keywords (`color`/`c`, `linestyle`/`ls`, `linewidth`/`lw`, `marker`, `markersize`/`ms`, `alpha`, `label`) are translated for plotly. Other keywords go to `plotly.graph_objects.Scatter`. Matplotlib format strings such as `'r--'` only work with matplotlib. The returned figure can then be changed with the plotly API.

```python
fig = ring_slot.plot_s_db(m=1, n=0, color='red', ls='--', lw=2, marker='o', backend='plotly')
fig.update_layout(title='Ring slot S21')
```

## Smith chart

The plotly Smith chart has the same options as the matplotlib one: `chart_type` (`'z'`, `'y'`, `'zy'`, `'yz'`), `r`, `draw_labels`, `ref_imm`, `draw_vswr` and `label_axes`.

```python
ring_slot.plot_s_smith(draw_labels=True, draw_vswr=[2])
```

```python
ring_slot.plot_s_smith(draw_labels=True, draw_vswr=[2], backend='plotly')
```

## Matplotlib-only functions

The global option doesn't affect these. They always use matplotlib:

* figure and style helpers in `skrf.plotting` such as `stylely`, `save_all_figs`, `func_on_all_figs`, `add_markers_to_lines` and `shade_bands`
* `NetworkSet.animate` and `NetworkSet.signature`
* the `Calibration`, `CalibrationSet`, `VectorFitting`, de-embedding and `Circuit.plot_graph` plots
* `Freespace.plot_ep`, `plot_mu` and `plot_ep_mu`

Those that accept a `backend` argument raise `NotImplementedError` for `backend='plotly'`.
